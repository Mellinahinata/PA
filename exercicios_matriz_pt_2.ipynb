```python
# 3 Utilize um for para exibir somente os números da primeira coluna da matriz.

matriz = [

[2, 4, 6],

[8, 10, 12],

[14, 16, 18]

]
for linha in matriz:
  print(linha[0])
```

```python
# 4 Altere o número 20 para 99 e exiba a matriz completa após a alteração.

matriz = [

[10, 20, 30],

[40, 50, 60]

]
matriz[0][1] = 99
print(matriz)
```

```python
# 6 crie uma lista vazia chamada coluna. Percorra a matriz, guarde nela os números da segunda coluna e, ao final, exiba a lista.

matriz = [

[3, 6, 9],

[12, 15, 18],

[21, 24, 27]

]
for linha in matriz:
  print(linha[1])
```

```python
# 7 Utilize dois comandos for para exibir todos os números da matriz, um por linha.

matriz = [

[1, 2],

[3, 4],

[5, 6]

]
for linha in matriz:
  for numero in linha:
    print(numero)
```

```python
# 8 crie um programa que some todos os números da segunda coluna e exiba o resultado.

matriz = [

[10, 20, 30],

[40, 50, 60],

[70, 80, 90]

]
soma = 0
for linha in matriz:
  soma += linha[1]
print(soma)
```

```python
# 9 Utilize dois comandos for para percorrer a matriz e encontrar o maior número. Ao final, exiba o resultado.

matriz = [

[12, 5, 8],

[20, 3, 15],

[7, 25, 10]

]
maior = matriz[0][0]
for linha in matriz:
  for numero in linha:
    if numero > maior:
      maior = numero
print(maior)
```

```python
# 10 Uma matriz armazena as notas de três alunos, sendo cada linha correspondente a um aluno.
#Faça um programa que:

#a) Exiba todas as notas do primeiro aluno.

#b) Exiba somente a segunda nota do segundo aluno.

#c) Utilize um for para exibir a terceira nota de cada aluno.

#d) Altere a primeira nota do terceiro aluno para 8 e exiba a matriz atualizada.

notas = [

[7, 8, 9],

[6, 5, 8],

[10, 9, 7]

]

print("a) Notas do primeiro aluno:")
print(notas[0])
print()
print("b) Segunda nota do segundo aluno:")
print(notas[1][1])
print()
print("c) Terceira nota de cada aluno:")
for aluno in notas:
  print(aluno[2])
print()
print("d) Matriz atualizada após alterar a primeira nota do terceiro aluno:")
notas[2][0] = 8
print(notas)
```
