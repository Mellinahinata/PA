# **Exercícios**

```python
# 1 – Nome em maiúsculo
# Crie um programa que solicite ao usuário seu nome e mostre esse nome completamente em letras maiúsculas.
nome = input("Digite seu nome: ")
print(nome.upper())
```

```python
# 2 – Nome em minúsculo
# Crie um programa que solicite o nome de uma cidade e exiba o nome todo em letras minúsculas
cidade = input("Digite o nome de uma cidade: ")
print(cidade.lower())
```

```python
# 3 – Primeira letra maiúscula
# Peça ao usuário para digitar seu nome completo e exiba o nome utilizando a primeira letra maiúscula
nome = input("Digite seu nome completo: ")
print(nome.title())
```

```python
# 4 – Organizando nomes
# Peça ao usuário para digitar uma frase com todas as letras minúsculas. Depois, mostre a frase com a primeira letra de cada palavra maiúscula.
frase = input("Digite uma frase com todas as letras minúsculas: ")
print(frase.title())
```

```python
# 5 – Substituindo palavras
# Peça ao usuário para digitar a frase:
# Eu gosto de estudar Java.
# Utilize o comando adequado para substituir a palavra Java por Python e mostre o resultado
frase = "Eu gosto de estudar Java"
print(frase.replace("Java", "Python"))
```

```python
# 6 – Removendo espaços
# Crie um programa que solicite ao usuário digitar seu nome com espaços antes e depois. Utilize o comando adequado para remover esses espaços e exiba o resultado.
nome = input("Digite seu nome com espaços antes e depois: ")
print(nome.strip())
```

```python
# 7 – Cadastro de aluno
# Crie um programa que solicite:
#  Nome do aluno;
#  Nome da escola;
#  Nome do curso.
# Depois, mostre:
#  Nome do aluno em letras maiúsculas;
#  Nome da escola em letras minúsculas;
#  Nome do curso com a primeira letra de cada palavra maiúscula.
nome = input("Digite o nome do aluno: ")
escola = input("Digite o nome da escola: ")
curso = input("Digite o nome do curso: ")
print(nome.upper())
print(escola.lower())
print(curso.title())
```

```python
# 8 – Corrigindo uma frase
# Peça ao usuário para digitar a frase:
# eu estou aprendendo java
# O programa deverá:
# 1. Substituir Java por Python.
# 2. Transformar a primeira letra de cada palavra em maiúscula.
# 3. Exibir a frase final
frase = "eu estou aprendendo java"
print(frase.replace("Java", "Python"))
print(frase.title())
```
