```python
# range()
for numero in range(10):
  print(numero)
```

```python
for numero in range(1,6):
  print(numero)
```

```python
for numero in range(0,11,2):
  print(numero)
```

```python
for numero in range(10,0,-1):
  print(numero)
```

```python
# match
numero = 2
match numero:
  case 1:
    print("você escolheu número 1")
  case 2:
    print("você escolheu número 2")
  case 3:
    print("você escolheu número 3")
```

```python
opcao = 1
match opcao:
  case 1:
    print("hambúrguer")
  case 2:
    print("pizza")
  case 3:
    print("hot-dog")
  case _:
    print("opção inválida")
```

```python
dia = "sábado"
match dia:
  case "segunda":
    print("dia de faltar")
  case "terça":
    print("aula da Letícia")
  case "quarta":
    print("metade da semana")
  case "quinta":
    print("perto de sexta")
  case "sexta":
    print("felicidade")
  case "sábado":
    print("enquanto não chega domingo")
  case "domingo":
    print("tristeza a noite")
  case _:
    print("opção inválida")
```
