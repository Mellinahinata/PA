```python
#soma
matriz = [
    [1,2],
    [3,4]
]
soma = 0
for linha in matriz:
  for numero in linha:
    soma = soma + numero
print("soma:", soma)
```

```python
#maior
matriz = [
    [8,3,12],
    [5,20,7]
]
maior = matriz[0][0]
for linha in matriz:
  for numero in linha:
    if numero > maior:
      maior = numero
print("maior número:", maior)
```
