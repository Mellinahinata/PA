# **Match**

```python
# 1 Crie um programa em Python que utilize a estrutura match case para verificar uma cor armazenada em uma variável. Se a cor for "azul", o programa deverá exibir a mensagem "Cor azul". Caso contrário, deverá mostrar "Outra cor".
cor = "azul"
match cor:
    case "azul":
        print("Cor azul")
    case _:
        print("Outra cor")
```

```python
# 2 Desenvolva um programa que utilize a estrutura match case para verificar o valor de uma variável chamada numero. O programa deverá apresentar o número por extenso, considerando as opções 1, 2 e 3.
numero = 2
match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
    case _:
        print("Número inválido")
```

```python
#3 Crie um programa que apresente uma opção de bebida de acordo com o número armazenado em uma variável. Utilize a estrutura match case para identificar as seguintes opções:

# · 1 – Água

# · 2 – Suco

# · 3 – Refrigerante

# Caso o número não corresponda a nenhuma das opções, o programa deverá exibir "Opção inválida".
bebida = 3
match bebida:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")
```

```python
# 4 Desenvolva um programa que utilize a estrutura match case para verificar se um dia da semana corresponde ao final de semana. Se o dia for "sábado" ou "domingo", deverá exibir "Final de semana". Para os demais dias, deverá mostrar "Dia de semana".
dia = "domingo"
match dia:
    case "sábado" | "domingo":
        print("Final de semana")
    case _:
        print("Dia de semana")
```

```python
# 5 Crie um programa que utilize a estrutura match case para verificar uma nota representada por uma letra. O programa deverá apresentar uma mensagem de acordo com as seguintes condições:

#· A – Excelente

#· B – Bom

#· C – Precisa melhorar

#Caso a letra informada não corresponda às opções, deverá exibir "Nota inválida".
nota = "g"
match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```

# **Range**

```python
# 1. Mostrar os números de 0 até 9
for i in range(10):
    print(i)
```

```python
# 2. Mostrar os números de 1 até 10
for i in range(1, 11):
    print(i)
```

```python
# 3. Mostrar os números de 5 até 15
for i in range(5, 16):
    print(i)
```

```python
# 4. Mostrar os números pares de 0 até 20
for i in range(0, 21, 2):
    print(i)
```

```python
# 5. Mostrar os números ímpares de 1 até 15
for i in range(1, 16, 2):
    print(i)
```

```python
# 6. Fazer uma contagem regressiva de 10 até 1
for i in range(10, 0, -1):
    print(i)
```

```python
# 7. Mostrar a tabuada do número 3
for i in range(1, 11):
    print(f"3 x {i} = {3 * i}")
```
