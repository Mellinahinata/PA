Compilador: traduz um código fonte de uma vez para código de máquina (ex: c/ctt)

interpretador: lê e escuta o código linha por linha sem gerar executável separado (ex:python)

SDK: conjunto de ferramentas, bibliotecas e documentação para desenvolver em uma plataforma (ex: android SDK)

IDE: ambiente completo com editor, debugger, terminal e recursos (ex: vs code, eclipse)

biblioteca: conjunto de funções prontas que você chama no código

framework: estrutura maior que define o “esqueleto” da aplicação e chama seu código

parâmetro: o valor que a função recebe

retorno: o valor que a função devolve
