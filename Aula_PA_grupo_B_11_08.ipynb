```python
# Exercício 7 – Classificação de idade
# Receba a idade de uma pessoa e classifique-a de acordo com sua faixa etária:
# Menor de 12 anos → Criança
# De 12 a 17 anos → Adolescente
# De 18 a 59 anos → Adulto
# 60 anos ou mais → Idoso
idade = int(input("digite sua idade: "))
if idade < 12:
  print("você é uma criança")
elif idade >= 12 and idade <= 17:
  print("você é um adolescente")
elif idade >= 18 and idade <= 59:
  print("você é um adulto")
else:
  print("você é um idoso")
```

```python
# Exercício 8 – Média de duas notas
# Receba duas notas, calcule a média e diga se o aluno foi aprovado ou reprovado. O aluno será aprovado se a média for maior ou igual a 7.
nota1 = float(input("digite a primeira nota: "))
nota2 = float(input("digite a segunda nota: "))
media = (nota1 + nota2) / 2
if media >= 7:
  print(f"você foi aprovado, sua média foi {media}")
else:
  print(f"você foi reprovado, sua média foi {media}")
```

```python
# Exercício 10 – Calculadora de IMC
# Receba o peso e a altura de uma pessoa, calcule o IMC usando a fórmula:
# IMC = peso / (altura × altura)
# Depois, classifique o resultado:
# IMC menor que 18.5 → Abaixo do peso
# IMC de 18.5 a 24.9 → Peso normal
# IMC de 25 a 29.9 → Sobrepeso
# IMC maior ou igual a 30 → Obesidade
peso = float(input("digite seu peso: "))
altura = float(input("digite sua altura: "))
imc = peso / (altura * altura)
if imc <= 18.5:
  print (f"seu imc é {imc},o peso está abaixo do normal")
elif imc <= 18.5 and imc >= 24.9:
  print (f"seu imc é {imc},o peso está normal")
elif imc <= 25 and imc >= 29.9:
  print (f"seu imc é {imc} peso está acima do normal")
elif imc >= 30:
  print (f"seu imc é {imc}, você está obeso 👀")
```
