```python
# Exercício 7 – Desafio simples
# Crie um programa que:
# • Guarde o nome de um aluno
# • Guarde duas notas
# • Calcule a média
# • Mostre uma frase assim: O aluno João tirou média 7.5

nome = str(input("digite o nome do aluno: "))
nota1 = float(input("digite a primeira nota: "))
nota2 = float(input("digite a segunda nota: "))

media = (nota1 + nota2) / 2

print(f"O aluno(a) {nome} tirou média {media}")
```
