```python
# Exercício 1 — Contagem

#Faça um programa que peça um número ao usuário e mostre todos os números de 1 até esse número usando while
numero = int(input("Digite um número: "))
contador = 1
while contador <= numero:
    print(contador)
    contador = contador + 1
```

```python
# Exercício 2 — Contagem regressiva

# Peça um número ao usuário e faça uma contagem até 0, mostrando cada número na tela.

#Exemplo:

#Digite um número: 5

#5

#4

#3

#2

#1

#0
numero = int(input("digite um número "))
while numero >= 0:
    print(numero)
    numero = numero - 1
```

```python
# Exercício 4 — Soma dos números
# Peça um número ao usuário e calcule a soma de todos os números de 1 até esse número
# Exemplo:
# Digite um número: 5
# Soma: 15
# Porque:
# 1 + 2 + 3 + 4 + 5 = 15
numero = int(input("digite um número: "))
soma = 0
contador = 1
while contador <= numero:
    soma = soma + contador
    contador = contador + 1
print("soma:", soma)
```

```python
# Exercício 5 — Tabuada
# Peça um número ao usuário e mostre a tabuada desse número de 1 até 10 usando while.
# Exemplo:
# Digite um número: 7
# 7 x 1 = 7
# 7 x 2 = 14
#...
#7 x 10 = 70
numero = int(input("digite um número para a tabuada: "))
contador = 1
while contador <= 10:
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
    contador = contador + 1
```

```python
# Exercício 6 — Senha
# Crie um programa que peça uma senha ao usuário.
# Enquanto a senha estiver incorreta, mostre:
# Senha incorreta!
# Quando o usuário digitar a senha correta, mostre:
# Acesso permitido!
# Considere que a senha correta seja "1234".
senha = input("digite a senha corretamente: ")
while senha != "1234":
    print("senha incorreta")
    senha = input("digite a senha de novo: ")
print("acesso permitido")
```

```python
# Exercício 7 — Somando até digitar zero
# Faça um programa que peça vários números ao usuário.
# O programa deve continuar pedindo números até que o usuário digite 0.
# Ao final, mostre a soma de todos os números digitados.
# Exemplo:
# Digite um número: 10
#Digite um número: 5
#Digite um número: 8
#Digite um número: 0
# Soma: 23
soma = 0
numero = int(input("digite um número: "))
while numero != 0:
    soma = soma + numero
    numero = int(input("digite um número: "))
print("soma:", soma)
```

```python
# Exercício 8 — Maior número
# Peça números ao usuário até que ele digite 0.
#Ao final, mostre qual foi o maior número digitado.
#Exemplo:
# Digite um número: 15
# Digite um número: 8
#Digite um número: 32
#Digite um número: 20
#Digite um número: 0
#Maior número: 32
maior = 0
numero = int(input("digite um número: "))
while numero != 0:
    if numero > maior:
        maior = numero
    numero = int(input("digite um número: "))
print("maior número:", maior)
```

```python
#Exercício 9 — Média das notas
#Faça um programa que peça 5 notas de um aluno usando while.
#Ao final, calcule e mostre a média das notas.
#Exemplo:
#Digite a nota: 7
#Digite a nota: 8
#Digite a nota: 6
#Digite a nota: 9
#Digite a nota: 10
#Média: 8.0
soma = 0
contador = 1
while contador <= 5:
    nota = float(input("digite a nota: "))
    soma = soma + nota
    contador = contador + 1
media = soma / 5
print("média:", media)
```

```python
# Exercício 10 — Caixa eletrônico
#Faça um programa que comece com um saldo de R$ 1000,00.
# Mostre um menu:
# 1 - Depositar
# 2 - Sacar
# 3 - Ver saldo
# 4 - Sair
saldo = 1000
while True:
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")
    opcao = int(input("digite a opção desejada: "))
    if opcao == 1:
        deposito = float(input("digite o valor do depósito: "))
        saldo = saldo + deposito
    elif opcao == 2:
        saque = float(input("digite o valor do saque: "))
        if saque > saldo:
            print("saldo insuficiente")
        else:
            saldo = saldo - saque
    elif opcao == 3:
        print("saldo:", saldo)
    elif opcao == 4:
        print("saindo...")
        break
    else:
        print("opção inválida")
```
